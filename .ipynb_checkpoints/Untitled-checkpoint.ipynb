```python
import matplotlib.pyplot as plt
import numpy as np

# 准备数据
metrics = ['Accuracy', 'Precision', 'F1-score', 'Specificity', 'Sensitivity']
values = [0.8273, 0.8117, 0.8389, 0.8321, 0.8221]

# 设置画布大小和分辨率
plt.figure(figsize=(8, 5), dpi=100)

# x轴位置
x_positions = np.arange(len(metrics))

# 绘制柱状图
plt.bar(x_positions, values, color='skyblue', width=0.6, edgecolor='black')

# 设置x轴刻度和标签
plt.xticks(x_positions, metrics, rotation=45, ha='right')

# y轴范围(可根据需要调整)
plt.ylim([0.0, 1.0])

# 添加数值标注
for i, v in enumerate(values):
    plt.text(i, v + 0.005, f"{v:.4f}", ha='center', color='black', fontsize=10)

# 添加标题和坐标轴标签
plt.title('Model Performance Metrics')
plt.ylabel('Score')

# 去掉上、右边框线条，使图表更美观
plt.box(False)

# 显示图表
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 准备数据
metrics = ['Accuracy', 'Precision', 'F1-score', 'Specificity', 'Sensitivity']

# 假设有三个模型的指标数据（根据实际情况进行替换）
model_names = ['Model A', 'Model B', 'Model C', 'Our Sentiment Model']
model_A_scores = [0.8273, 0.8117, 0.8389, 0.8321, 0.8221]
model_B_scores = [0.9350, 0.9101, 0.9435, 0.8360, 0.8290]
model_C_scores = [0.7990, 0.7750, 0.8300, 0.6255, 0.7155]
Our_model(SCB) = [0.8534, 0.8917, 0.8200, 0.8255, 0.8055]

all_scores = [model_A_scores, model_B_scores, model_C_scores, Our_model]

# 每个指标对应的x轴坐标
x_positions = np.arange(len(metrics))

# 每组柱子的总数（即模型数）
num_models = len(model_names)

# 柱子宽度
bar_width = 0.2

# 设置画布大小和分辨率
plt.figure(figsize=(10, 6), dpi=100)

# 对每个模型绘制柱状图
for i, scores in enumerate(all_scores):
    # 柱子在x方向的位置偏移，以使不同模型分组排列
    offset = (i - (num_models - 1) / 2) * bar_width
    plt.bar(x_positions + offset, scores, width=bar_width, label=model_names[i], edgecolor='black')

# 设置x轴刻度和标签
plt.xticks(x_positions, metrics, rotation=45, ha='right')

# 设置y轴范围(可根据需要调整)
plt.ylim([0.0, 1.0])

# 添加标题和坐标轴标签
plt.title('Comparison of Model Performance Metrics')
plt.ylabel('Score')

# 添加图例
plt.legend()

# 添加数值标注（可选）
for i, scores in enumerate(all_scores):
    offset = (i - (num_models - 1) / 2) * bar_width
    for j, v in enumerate(scores):
        plt.text(x_positions[j] + offset, v + 0.005, f"{v:.4f}", ha='center', color='black', fontsize=9, rotation=90)

plt.tight_layout()
plt.show()
```
